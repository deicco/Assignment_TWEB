```python
ax = df_actors['id'].value_counts() \
    .head(10) \
    .plot(kind = 'bar', title = 'Most actors id for each film')
ax.set_xlabel('id actors')
ax.set_ylabel('numbers of id')
```

```python
ax = df_actors['name'].value_counts() \
    .head(10) \
    .plot(kind = 'bar', title = 'Most actors names for each film')
ax.set_xlabel('actors name')
ax.set_ylabel('numbers of rows')
```

```python
ax = df_actors['role'].value_counts() \
    .head(10) \
    .plot(kind = 'bar', title = 'Most actors roles for each film')
ax.set_xlabel('actors role')
ax.set_ylabel('numbers of rows')
```

Starting data cleaning

Drop duplicates rows

```python
df_actors = df_actors.drop_duplicates()
```

Drop rows with null or missing values

```python
df_actors = df_actors.dropna()
```

Drop any whitespace from the column names

```python
df_actors = df_actors.columns.str.strip()
```
